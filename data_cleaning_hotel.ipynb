```python
import pandas as pd
```

```python
df = pd.read_csv('hotel_bookings.csv')
```

```python
df.info()
df.describe()
df.head()
```

```python
df.columns
```

```python
# Rename columns
df.rename({'adults': 'num_adults', 'children': 'num_children', 'babies': 'num_babies'}, axis=1, inplace=True)
df.head(10)
```

```python
#NaNs
df.isnull().sum() * 100 / len(df)
```

```python
# Dealing with NaNs/Null Vules
df['agent'].unique()
df[df['agent'] == 5]
df['agent'].fillna(-1, inplace=True)
df[df['num_children'].isna()]
df[df['country'].isna()][['hotel', 'is_canceled', 'country']]

#Replace NaN with 'Unknown'
df['country'].fillna('Unknown', inplace=True)

# Drop num_children rows with NaN values
df.dropna(subset=['num_children'], inplace=True)

# Drop columns of company
df.drop(columns=['company'], inplace=True)
```

```python
# Check column data types
df.dtypes
```

```python
df = df.astype({'is_canceled': 'boolean', 'is_repeated_guest': 'boolean', 'num_children': 'int64', 'agent': 'int64'})
df.dtypes
```

```python
#Bin  columns
df['lead_time'].unique()
df['lead_time'].describe()

bins = [0, 100, 200, 300, 400, 500, 600, 700, 800]
labels = ['0-100', '101-200', '201-300', '301-400', '401-500', '510-600', '601-700', '701-800']

df['lead_time_binned'] = pd.cut(df['lead_time'], bins=bins, labels=labels)

df[['lead_time', 'lead_time_binned']]
```
